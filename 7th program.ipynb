```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Category': ['A', 'A', 'B', 'B'],
    'Data': [10, 20, 30, 40]
})
```

```python
print("Original DataFrame:")
print(df)
```

```python
grouped = df.groupby('Category')
```

```python
print("\nSum of Data by Category:")
print(grouped['Data'].sum())
```

```python
print("\nMean and Standard Deviation:")
print(grouped['Data'].agg(['mean', 'std']))
```
