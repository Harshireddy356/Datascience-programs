```python
import pandas as pd
```

```python
import numpy as np
```

```python
data = pd.Series(
    np.random.randn(6),
    index=[['a', 'a', 'b', 'b', 'c', 'c'],
           [1, 2, 1, 2, 1, 2]]
)
```

```python
print('Original Series:')
print(data)
```

```python
print('\nValues for index b:')
print(data['b'])
```

```python
print('\nValues with second index = 1:')
print(data[:, 1])
```

```python
df = data.unstack()
```

```python
print('\nUnstacked DataFrame:')
print(df)
```

```python
print('\nStacked Series:')
print(df.stack())
```

C. Merge and Combine

```python
df1 = pd.DataFrame({'a': [1, np.nan, 3], 'b': [4, 5, 6]})
```

```python
df2 = pd.DataFrame({'a': [7, 8, 9], 'b': [np.nan, 11, 12]})
```

```python
merged = df1.combine_first(df2)
```

```python
print('\nMerged DataFrame:')
print(merged)
```
